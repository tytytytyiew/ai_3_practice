## 1. Загрузка реальных данных
Используется UCI Adult Census Income — реальный набор переписных данных. В нём встречаются пропуски, числовые и категориальные признаки. Для загрузки нужен интернет.

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from pathlib import Path

DATA_PATH = Path("../data/practice2_retail_eda.csv")
df = pd.read_csv(DATA_PATH)
print("Источник: локальный учебный CSV, синтетические данные с внесёнными пропусками и выбросами")
print("Размер:", df.shape)
display(df.head())
```

## 2. Общий осмотр и типы признаков

```python
print(df.info())
print("\\nПропуски:")
display(df.isna().sum().sort_values(ascending=False).to_frame("missing"))
print("\\nСтатистики числовых признаков:")
display(df.describe().T)
numeric_cols = df.select_dtypes(include=np.number).columns.tolist()
categorical_cols = df.select_dtypes(exclude=np.number).columns.tolist()
print("Числовые признаки:", numeric_cols)
print("Категориальные признаки:", categorical_cols)
```

## 3. Распределения и выбросы

```python
for col in ["age", "monthly_spend", "visits_per_month"]:
    plt.figure(figsize=(7, 4))
    plt.hist(df[col].dropna(), bins=30, edgecolor="black")
    plt.title(f"Распределение: {col}")
    plt.xlabel(col)
    plt.ylabel("Частота")
    plt.tight_layout()
    plt.show()

plt.figure(figsize=(7, 4))
plt.boxplot(df["monthly_spend"].dropna(), vert=False)
plt.title("Boxplot: monthly_spend")
plt.xlabel("Расходы за месяц")
plt.tight_layout()
plt.show()

df["product_category"].value_counts().head(12).plot(kind="bar", figsize=(9, 4))
plt.title("Частоты категорий товаров")
plt.xlabel("Категория")
plt.ylabel("Количество")
plt.xticks(rotation=45, ha="right")
plt.tight_layout()
plt.show()
```

## 4. Корреляции и целевая переменная

```python
df["returned"].value_counts(normalize=True).mul(100).plot(kind="bar", figsize=(6, 4))
plt.title("Доля возвращённых заказов")
plt.ylabel("Доля, %")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```
